# Explainable Fake-News Detection - demo notebook
Run from VS Code (select the `.venv` kernel). Switch `DATASET` to `'kaggle'` or `'liar'` after downloading the data.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))   # make the `fakenews` package importable

from fakenews.data import load_dataset, split_dataset
from fakenews.classic import build_pipeline
from fakenews.evaluate import compute_metrics
from fakenews.explain import explain, plot_contributions

DATASET = 'sample'
```

```python
df = load_dataset(DATASET)
train, val, test = split_dataset(df)
print(len(df), 'articles |', df['label'].value_counts().to_dict(), '(0=REAL, 1=FAKE)')
df.sample(5, random_state=1)
```

```python
model = build_pipeline('lr').fit(train['text'], train['label'])
preds = model.predict_proba(test['text'].tolist()).argmax(1)
compute_metrics(test['label'], preds)
```

## Explain a prediction
`occlusion` needs no extra libraries. Use `'lime'` or `'shap'` after `pip install lime shap`.

```python
text = 'The government has announced a new tax cut for all citizens.'
print(model.predict_proba([text])[0])   # [P(REAL), P(FAKE)]
pairs = explain('occlusion', model.predict_proba, text, top_k=8)
pairs
```

```python
plot_contributions(pairs)
```
